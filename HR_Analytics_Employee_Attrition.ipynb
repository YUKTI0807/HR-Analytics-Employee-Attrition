# 4. Exploratory Data Analysis: Exploring the data to find patterns, relationships, and useful information.

## 4.1 Attrition Overview

### Count employees who left/stayed

```python
df["Attrition"].value_counts()
```

### Percentage

```python
df["Attrition"].value_counts(normalize=True) * 100
```

### Calculate attrition rate

```python
attrition_rate = (df["Attrition"] == "Yes").mean() * 100
# print("Attrition Rate: ",round(attrition_rate,2), "%")
print(f"Attrition Rate: {attrition_rate : .2f} %")
```

```python
df["Attrition"].value_counts().plot(kind="bar")
plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

## 4.2 Department Analysis

### Does attrition differ between departments?

```python
df["Department"].value_counts()
```

```python
pd.crosstab(df["Department"], df["Attrition"])
```

### Calculate attrition rate

```python
department_attrition = pd.crosstab(df["Department"], df["Attrition"],normalize="index") * 100
department_attrition
```

### Visualize

```python
department_attrition["Yes"].plot(kind="bar")

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

## 4.3 Age Analysis

### Is attrition different for different age groups?

```python
df["Age"].describe()
```

### Create age groups:

```python
bins = [0,24,34,44,54,100]

labels = ["Under 25", "25-34", "35-44", "45-54", "55+"]

df["AgeGroup"] = pd.cut(df["Age"],bins=bins, labels=labels)
```

```python
df["AgeGroup"].value_counts().sort_index()
```

```python
age_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

age_attrition
```

```python
age_attrition["Yes"].plot(kind="bar")

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

## 4.4 Salary Analysis

### Is employee salary associated with attrition?

```python
df["MonthlyIncome"].describe()
```

### Create salary groups

```python
df["SalaryGroup"] = pd.qcut(
    df["MonthlyIncome"],
    q=3,
    labels=["Low", "Medium", "High"]
)
```

```python
df["SalaryGroup"].value_counts()
```

### Calculate attrition

```python
salary_attrition = pd.crosstab(
    df["SalaryGroup"],
    df["Attrition"],
    normalize="index"
) * 100

salary_attrition
```

### Visualize

```python
salary_attrition["Yes"].plot(kind="bar")

plt.title("Attrition Rate by Salary Group")
plt.xlabel("Salary Group")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

## 4.5 Job Role Analysis

### Does attrition differ across job roles?

```python
df["JobRole"].value_counts()
```

### Calculate attrition

```python
role_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

role_attrition
```

### Visualize

```python
role_attrition["Yes"].sort_values().plot(kind="barh")

plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")
plt.show()

# This is an important analysis because HR can see which roles have different attrition levels.
```

## 4.6 Years at Company

### Does the number of years an employee has worked at the company relate to attrition?

```python
df["YearsAtCompany"].describe()
```

### Create groups

```python
bins = [-1, 2, 5, 10, 100]

labels = [
    "0-2 Years",
    "3-5 Years",
    "6-10 Years",
    "10+ Years"
]

df["YearsGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=bins,
    labels=labels
)
```

```python
years_attrition = pd.crosstab(
    df["YearsGroup"],
    df["Attrition"],
    normalize="index"
) * 100

years_attrition
```

### Visualize

```python
years_attrition["Yes"].plot(kind="barh")

plt.title("Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")
plt.show()
```
